```python
import os
from glob import glob
import numpy as np
import matplotlib.pyplot as plt

os.makedirs('img', exist_ok=True)
os.getcwd()
```

```python

outname="freefiles/freefile"
nstlim=400 # ps
n_hills=500
timestep=0.001 #fs -- 0.001
```

```python
data = np.loadtxt(f'{outname}.dat', comments="#")
print('data shape: ', data.shape)

rc = data[:,0] 
fes = data[:,1] 

plt.plot(rc, fes)
plt.xlabel('r1 - r2 (Å)', fontsize=12)
plt.ylabel('Free Energy (kcal/mol)', fontsize=12)
plt.savefig('img/fes.png', dpi=300)
```

```python
fnames = sorted(glob(f'{outname}_timedep?.dat'))
print('found files: ', fnames)

ii = nstlim / len(fnames)
for i,fname in enumerate(fnames):
    data = np.loadtxt(f'{fname}', comments="#")
    print('data shape: ', data.shape)

    rc = data[:,0] 
    fes = data[:,1] 

    label = f'{ii} ps'

    plt.plot(rc, fes, label=label)
    plt.xlabel('r1 - r2 (Å)', fontsize=12)
    plt.ylabel('Free Energy (kcal/mol)', fontsize=12)
    plt.legend(loc=2, ncols=3)
    plt.ylim(np.round(fes.min())-5, np.round(fes.max())+5)

    ii=ii + (nstlim/len(fnames))

plt.savefig('img/fes_timedep.png', dpi=300)
```
